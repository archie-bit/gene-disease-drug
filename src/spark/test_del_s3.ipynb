```python
import boto3
# ... (Your existing imports and config remain the same) ...

# MinIO Config
MINIO_ACCESS_KEY = "minioadmin"
MINIO_SECRET_KEY = "minioadmin"
MINIO_ENDPOINT = "http://minio:9000"

def delete_bucket_and_contents(bucket_name):
    """Deletes all objects in a bucket and then deletes the bucket itself."""
    
    # Initialize the S3 client/resource
    s3 = boto3.resource('s3',
                        endpoint_url=MINIO_ENDPOINT,
                        aws_access_key_id=MINIO_ACCESS_KEY,
                        aws_secret_access_key=MINIO_SECRET_KEY,
                        config=boto3.session.Config(signature_version='s3v4'))
    
    bucket = s3.Bucket(bucket_name)

    if bucket in s3.buckets.all():
        print(f"Attempting to delete objects in bucket: {bucket_name}...")
        
        # 1. Delete all objects (and object versions, if versioning is enabled)
        try:
            # Use the Collection.delete() method on the objects collection
            bucket.objects.all().delete()
            print(f"Successfully deleted all objects in {bucket_name}.")
            
            # 2. Delete the bucket
            bucket.delete()
            print(f"Successfully deleted bucket: {bucket_name}.")
        except Exception as e:
            print(f"Error during bucket deletion for {bucket_name}: {e}")
    else:
        print(f"Bucket {bucket_name} does not exist.")

# --- To use this function ---
# Before stopping Spark, or as a separate step:
# delete_bucket_and_contents("datalake")
delete_bucket_and_contents("datalake")
```
